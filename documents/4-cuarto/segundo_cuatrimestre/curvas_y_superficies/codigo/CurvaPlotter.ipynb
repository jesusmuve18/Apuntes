# Dibujar curva en el espacio

Programa que dibuja curvas

## Definición de funciones

#### Fundamento matemático

Usaremos las siguientes propiedades para curvas planas:
$$k_\alpha(s) = \frac{det(\alpha'(s), \alpha''(s))}{|\alpha'(s)|^3} $$
Y las siguientes propiedades para curvas en el espacio:
$$k_\alpha(s) = \frac{|\alpha'(s) \wedge \alpha''(s)|}{|\alpha'(s)|^3} \ \ \ \ \ \ \ \ \tau_\alpha = - \frac{det(\alpha'(s), \alpha''(s), \alpha'''(s))}{|\alpha'(s) \wedge \alpha''(s)|^2}$$

Además, para facilitar el código usaremos el siguiente desarrollo por la primera fila:
\begin{align*}
  (x', y', z') \wedge (x'', y'', z'') =
  \begin{vmatrix}
    i & j & k\\
    x' & y' & z'\\
    x'' & y'' & z''
  \end{vmatrix} =
  i \begin{vmatrix}
    y' & z'\\
    y'' & z''
  \end{vmatrix}
  -j \begin{vmatrix}
    x' & z'\\
    x'' & z''
  \end{vmatrix}
  +k \begin{vmatrix}
    x' & y'\\
    x'' & y''
  \end{vmatrix}
\end{align*}
y el siguiente desarrollo para reutilizar el cálculo anterior:
\begin{align*}
  \begin{vmatrix}
    x' & y' & z'\\
    x'' & y'' & z''\\
    x''' & y''' & z'''
  \end{vmatrix} =
  x'''\begin{vmatrix}
    y' & z'\\
    y'' & z''
  \end{vmatrix}
  -y''' \begin{vmatrix}
    x' & z'\\
    x'' & z''
  \end{vmatrix}
  +z''' \begin{vmatrix}
    x' & y'\\
    x'' & y''
  \end{vmatrix}
\end{align*}

#### Código

```python
import plotly.graph_objects as go                 # Representación de funciones
import numpy as np                                # Matemáticas
import re                                         # Expresiones regulares
from scipy.integrate import cumulative_trapezoid  # Integración acumulativa
from scipy.integrate import solve_ivp             # Resolución de PVI (Initial Values Problem)

######################################################################################################################

def crear_f(eq):
  # Traducción de funciones de texto plano a un lenguaje que python entienda

  sustituciones = {
    '^': '**',
    'sen': 'sin', 'tg' : 'tan',
    'ln' : 'log',
    'arcsen' : 'arcsin', 'arctg' : 'arctan',
    'asin' : 'arcsin', 'asen' : 'arcsin',
    'acos' : 'arccos',
    'atan' : 'arctan', 'atg' : 'arctan'
  }

  entorno_math = {
    'sin': np.sin, 'cos': np.cos, 'tan': np.tan, 'tg' : np.tan,
    'arcsin': np.arcsin, 'arccos': np.arccos, 'arctan': np.arctan,
    'exp': np.exp, 'log': np.log, 'log10': np.log10,
    'sqrt': np.sqrt, 'abs': np.abs,
    'pi': np.pi, 'e': np.e
  }

  f = []

  # Iteramos sobre cada cadena de ecuación en la lista 'eq'
  for expresion in eq:
    # Manejo de las sustituciones
    for viejo, nuevo in sustituciones.items():
        expresion = expresion.replace(viejo, nuevo)

    # Manejo de las yuxtaposiciones
    texto_eq = re.sub(r'(\d)([a-zA-Z])', r'\1*\2', expresion)

    # Creación de la función lambda.
    f_lambda = lambda s, eq_procesada=texto_eq: eval(eq_procesada, entorno_math, {"s": s})

    f.append(f_lambda)

  return f

######################################################################################################################

def dibujarCurva2d(f, s_interval=[-10,10], n=300, color='darkblue', proportion="false"):
  # Calculo las coordenadas de cada punto
  s = np.linspace(s_interval[0], s_interval[1], n)
  x = f[0](s)
  y = f[1](s)

  # Por si alguna ecuación es un valor fijo
  if isinstance(x, (int, float)): x = np.full_like(s, x)
  if isinstance(y, (int, float)): y = np.full_like(s, y)

  # Calculo la curvatura
  k = curvatura2D(x, y, s)
  customdata = np.stack([k], axis=-1)

  # Crear el gráfico 2D usando go.Scatter
  fig = go.Figure(data=go.Scatter(
    x=x, y=y,
    mode='lines',
    line=dict(
      color=color,
      width=4
    ),
    customdata=customdata,
    hovertemplate=(
      "x(s): %{x:.2f}<br>" +
      "y(s): %{y:.2f}<br>" +
      "──────────────<br>" +
      "k(s): %{customdata[0]:.4f}" +
      "<extra></extra>" # Esto oculta el recuadro secundario con el nombre de la traza
    )
  ))

  # Configurar el diseño
  fig.update_layout(
    margin=dict(l=20, r=20, b=20, t=40),
    title_x=0.5,
    # Esto hace que 1 unidad en X mida lo mismo que 1 unidad en Y
    yaxis=dict(scaleanchor="x", scaleratio=1)
  )

  # Mostrar el gráfico
  fig.show()

######################################################################################################################

def dibujarCurva3d(f, s_interval=[-10,10], n=300, color='darkblue', proportion="false"):
  # Calculo las coordenadas de cada punto
  s = np.linspace(s_interval[0], s_interval[1], n)
  x = f[0](s)
  y = f[1](s)
  z = f[2](s)

  # Por si alguna ecuación es un valor fijo
  if isinstance(x, (int, float)): x = np.full_like(s, x)
  if isinstance(y, (int, float)): y = np.full_like(s, y)
  if isinstance(z, (int, float)): z = np.full_like(s, z)

  # Calculo la curvatura y la torsion
  k, t = curvaturaTorsion3D(x, y, z, s)
  customdata = np.stack([k, t], axis=-1)

  # Crear el gráfico
  fig = go.Figure(data=go.Scatter3d(
    x=x, y=y, z=z,
    mode='lines',
    line=dict(
      color=color,
      width=4
    ),
    customdata=customdata,
    hovertemplate = (
      "x(s): %{x:.2f}<br>" +
      "y(s): %{y:.2f}<br>" +
      "z(s): %{z:.2f}<br>" +
      "──────────────<br>" +
      "k(s): %{customdata[0]:.4f}<br>" +
      "τ(s): %{customdata[1]:.4f}"+
      "<extra></extra>"
    )
  ))

  # Configurar el diseño
  fig.update_layout(
    margin=dict(l=0, r=0, b=0, t=0),
    scene=dict(aspectmode= 'data' if proportion else 'auto') if proportion else dict(),
    title_x=0.5,
  )

  # Mostrar el gráfico
  fig.show()

######################################################################################################################

def dibujarCurva2dCurvatura(k, a=0, s_interval=[-10, 10], n=300, color="darkblue", proportion="false"):
  # Definición del dominio
  s = np.linspace(s_interval[0], s_interval[1], int(n))

  # Manejo de los grados de libertad (si se pone solo un número se crear un vector 3-dimensional con dicho valor)
  if np.isscalar(a): a = [a, a, a]

  k_vals = k(s)

  # Por si es un valor fijo
  if isinstance(k_vals, (int, float)): k_vals = np.full_like(s, k_vals)

  # Integración acumulativa
  theta_vals = cumulative_trapezoid(k_vals, s, initial=0)

  # Ajustamos theta para que en s = a[0] valga 0 (Teorema Fundamental)
  theta_vals -= np.interp(a[0], s, theta_vals)  # Trasladamos para imponer theta(a[0])=0

  # Calculamos alpha (en cada una de sus componentes)
  alpha_x_vals = cumulative_trapezoid(np.cos(theta_vals), s, initial=0)
  alpha_x_vals -= np.interp(a[1], s, alpha_x_vals)  # Trasladamos para imponer alpha_x(a[1])=0

  alpha_y_vals = cumulative_trapezoid(np.sin(theta_vals), s, initial=0)
  alpha_y_vals -= np.interp(a[2], s, alpha_y_vals)  # Trasladamos para imponer alpha_y(a[2])=0

  # Dibujamos la curva
  dibujarCurva2d([lambda _: alpha_x_vals, lambda _: alpha_y_vals], s_interval=s_interval, n=n, color=color, proportion=proportion)

######################################################################################################################

def dibujarCurva3dCurvatura(k, tau, a=0, s_interval=[-10, 10], n=300, color="darkblue", proportion="false"):
  # Definición del dominio
  s_eval = np.linspace(s_interval[0], s_interval[1], int(n))

  # Manejo de curvatura y torsión constantes
  k_func = (lambda s: k) if isinstance(k, (int, float)) else k
  tau_func = (lambda s: tau) if isinstance(tau, (int, float)) else tau

  # COMPROBACIÓN DE CURVATURA k>0
  k_vals = k_func(s_eval)
  if np.isscalar(k_vals): k_vals = np.full_like(s_eval, k_vals)
  if np.any(k_vals <= 1e-7): raise ValueError("Error Matemático: La curvatura k(s) se hace nula o negativa en el dominio seleccionado. El triedro de Frenet no está definido en ese punto.")

  # Manejo de los grados de libertad (si se pone solo un número se crear un vector 3-dimensional con dicho valor)
  if np.isscalar(a): a = [a, a, a]

  # Planteamos el sistema de EDOs
  def EDO(s, Y):
    t = Y[0:3]
    n = Y[3:6]
    b = Y[6:9]

    # Ecuaciones diferenciales
    dt = k_func(s) * n
    dn = -k_func(s) * t - tau_func(s) * b
    db = tau_func(s) * n

    return np.concatenate((dt, dn, db))

  # Condiciones iniciales (elegidas para ser una base ortonormal)
  t_0 = np.array([1.0, 0.0, 0.0])
  n_0 = np.array([0.0, 1.0, 0.0])
  b_0 = np.array([0.0, 0.0, 1.0])

  Y0 = np.concatenate((t_0, n_0, b_0))

  # Resolvemos la EDO
  sol = solve_ivp(EDO, [s_interval[0], s_interval[1]], Y0, t_eval=s_eval, method='RK45') # Utilizamos el método Runge-Kutta

  # Extraemos el vector tangente t(s)
  t_x = sol.y[0]
  t_y = sol.y[1]
  t_z = sol.y[2]

  # Calculamos alpha (en cada una de sus componentes)
  alpha_x = cumulative_trapezoid(t_x, s_eval, initial=0)
  alpha_x -= np.interp(a[0], s_eval, alpha_x) # Trasladamos para imponer alpha_x(a[0])=0

  alpha_y = cumulative_trapezoid(t_y, s_eval, initial=0)
  alpha_y -= np.interp(a[1], s_eval, alpha_y) # Trasladamos para imponer alpha_x(a[1])=0

  alpha_z = cumulative_trapezoid(t_z, s_eval, initial=0)
  alpha_z -= np.interp(a[2], s_eval, alpha_z) # Trasladamos para imponer alpha_x(a[2])=0

  # Dibujamos la curva
  dibujarCurva3d([lambda _: alpha_x, lambda _: alpha_y, lambda _: alpha_z], s_interval=s_interval, n=n, color=color, proportion=proportion)

######################################################################################################################

def curvatura2D(x, y, s):
  # Primera derivada (x', y')
  dx = np.gradient(x, s)
  dy = np.gradient(y, s)

  # Segunda derivada (x'', y'')
  ddx = np.gradient(dx, s)
  ddy = np.gradient(dy, s)

  # Aplicamos la fórmula de la curvatura
  numerador = dx * ddy - dy * ddx   # det(\alpha', \alpha'')
  denominador = (dx**2 + dy**2)**1.5

  # Usamos np.divide con 'out' y 'where' para evitar divisiones por cero en puntos donde la curva se detenga o haya singularidades.
  k = np.divide(numerador, denominador, out=np.full_like(numerador, np.nan, dtype=np.float64), where=denominador!=0)

  return k

######################################################################################################################

def curvaturaTorsion3D(x, y, z, s):
  # Primera derivada (x', y', z')
  dx = np.gradient(x, s)
  dy = np.gradient(y, s)
  dz = np.gradient(z, s)

  # Segunda derivada (x'', y'', z'')
  ddx = np.gradient(dx, s)
  ddy = np.gradient(dy, s)
  ddz = np.gradient(dz, s)

  # Tercera derivada (x''', y''', z''')
  dddx = np.gradient(ddx, s)
  dddy = np.gradient(ddy, s)
  dddz = np.gradient(ddz, s)

  # Producto vectorial \alpha' \wedge \alpha''
  prod_vec_x = dy * ddz - dz * ddy
  prod_vec_y = dx * ddz - dz * ddx
  prod_vec_z = dx * ddy - dy * ddx

  # Módulo de la primera derivada, |\alpha'|
  mod_d = np.sqrt(dx**2 + dy**2 + dz**2)

  # Módulo del producto vectorial |\alpha' \wedge \alpha''|
  mod_prod_vec = np.sqrt(prod_vec_x**2 + prod_vec_y**2 + prod_vec_z**2)

  # --- CÁLCULO DE LA CURVATURA (k) ---
  denominador_k = mod_d**3
  k = np.divide(mod_prod_vec, denominador_k, out=np.zeros_like(mod_prod_vec), where=denominador_k!=0)

  # --- CÁLCULO DE LA TORSIÓN (t) ---
  numerador_t = - dddx * prod_vec_x + dddy * prod_vec_y - dddz * prod_vec_z # Determinante desarrollado previamente
  denominador_t = mod_prod_vec**2
  t = np.divide(numerador_t, denominador_t, out=np.full_like(numerador_t, np.nan, dtype=np.float64), where=denominador_t!=0)

  return k, t
```

## Uso de funciones 2D

### Dibujar curva 2D a partir de ecuaciones paramétricas

A continuación se muestra un panel con los siguientes controles:
- `x`: Componente $x$ de la curva
- `y`: Componente $y$ de la curva
- `s_0`: Inicio del intervalo del dominio
- `s_1`: Final del intervalo del dominio
- `n`: Número de puntos para la representación (evitar picos)

De esta forma, una curva $\alpha$ estará definida como
\begin{align}
  \alpha : [s_0, s_1] & \to \mathbb{R}^2\\
  s & \mapsto (x(s), y(s))
\end{align}

---
*Para dibujar la curva (si no lo hace automáticamente tras cambiar los parámetros) se podrá pulsar `Alt+Enter` o pulsar en el icono ▶ de la izquierda de la celda*

```python
# --- FORMULARIO NATIVO DE COLAB ---
# @title {"run":"auto"}
# @markdown ### Ecuaciones Paramétricas
x = "cos(s)*ln(s)" # @param {type:"string", "placeholder":"x(s)"}
y = "sen(s)" # @param {type:"string", "placeholder":"y(s)"}
# @markdown ### Dominio
s_0 = 0.01 # @param {type:"slider", min:-200, max:200, step:0.01}
s_1 = 200 # @param {type:"slider", min:-200, max:200, step:0.01}
# @markdown ### Parámetros de visualización
n = 10000 # @param {type:"slider", min:0, max:10000, step:5}
color = "purple" # @param ["darkblue", "red", "forestgreen", "darkorange", "purple"]
# @markdown
mantener_proporcion = True # @param {"type":"boolean"}

try:
    # Convertimos el texto a matemáticas
    f = crear_f([x, y])

    dibujarCurva2d(f=f, s_interval=[s_0, s_1], n=n, color=color, proportion=mantener_proporcion)

except Exception as e:
    print(f"Error en la entrada: {e}")
```

### Dibujar curva 2D a partir de curvatura

A continuación se muestra un panel con los siguientes controles:

De esta forma, una curva $\alpha$ estará definida como $\alpha : [s_0, s_1] \to \mathbb{R}^2$ tal que $k_\alpha(s) = k(s)$ para todo $s\in [s_0, s_1]$

---
*Para dibujar la curva (si no lo hace automáticamente tras cambiar los parámetros) se podrá pulsar `Alt+Enter` o pulsar en el icono ▶ de la izquierda de la celda*

```python
# --- FORMULARIO NATIVO DE COLAB ---
# @title {"run":"auto"}
# @markdown ### Curvatura
k = "s^2" # @param {type:"string", "placeholder":"k(s)"}
# @markdown ### Grados de libertad
a_x = 0 # @param {type:"number", "placeholder":"a_x"}
a_y = 0 # @param {type:"number", "placeholder":"a_y"}
a_z = -1.58 # @param {type:"number", "placeholder":"a_z"}
# @markdown ### Dominio
s_0 = -27.77 # @param {type:"slider", min:-200, max:200, step:0.01}
s_1 = 22.42 # @param {type:"slider", min:-200, max:200, step:0.01}
# @markdown ### Parámetros de visualización
n = 10000 # @param {type:"slider", min:0, max:10000, step:5}
color = "darkblue" # @param ["darkblue", "red", "forestgreen", "darkorange", "purple"]
# @markdown
mantener_proporcion = True # @param {"type":"boolean"}

try:
    dibujarCurva2dCurvatura(k=crear_f([k])[0], a=[a_x, a_y, a_z] , s_interval=[s_0, s_1], n=n, color=color, proportion=mantener_proporcion)

except Exception as e:
    print(f"Error en la entrada: {e}")
```

### Dibujar curva 3D a partir de curvatura y torsión

A continuación se muestra un panel con los siguientes controles:
- `k`: Función de curvatura
- `tau`: Función de torsión
- `a`: Punto de $R^3$
- `s_0`: Inicio del intervalo del dominio
- `s_1`: Final del intervalo del dominio
- `n`: Número de puntos para la representación (evitar picos)

De esta forma, una curva $\alpha$ estará definida como $\alpha : [s_0, s_1] \to \mathbb{R}^2$ tal que $k_\alpha(s) = k(s)$, $\tau_\alpha(s) = \tau(s)$ para todo $s\in [s_0, s_1]$. Además, se verificará que \alpha(a) = (0,0,0).

---
*Para dibujar la curva (si no lo hace automáticamente tras cambiar los parámetros) se podrá pulsar `Alt+Enter` o pulsar en el icono ▶ de la izquierda de la celda*

```python
# --- FORMULARIO NATIVO DE COLAB ---
# @title {"run":"auto"}
# @markdown ### Curvatura y Torsión
k = "ln(s)" # @param {type:"string", "placeholder":"k(s)"}
tau = "ln(s)" # @param {type:"string", "placeholder":"tau(s)"}
# @markdown ### Grados de libertad
a_x = 0 # @param {type:"number", "placeholder":"a_x"}
a_y = 0 # @param {type:"number", "placeholder":"a_y"}
a_z = 0 # @param {type:"number", "placeholder":"a_z"}
# @markdown ### Dominio
s_0 = 1.55 # @param {type:"slider", min:-200, max:200, step:0.01}
s_1 = 200 # @param {type:"slider", min:-200, max:200, step:0.01}
# @markdown ### Parámetros de visualización
n = 10000 # @param {type:"slider", min:0, max:10000, step:5}
color = "red" # @param ["darkblue", "red", "forestgreen", "darkorange", "purple"]
# @markdown
mantener_proporcion = False # @param {"type":"boolean"}

try:
    dibujarCurva3dCurvatura(k=crear_f([k])[0],tau=crear_f([tau])[0], a=[a_x, a_y, a_z] , s_interval=[s_0, s_1], n=n, color=color, proportion=mantener_proporcion)

except Exception as e:
    print(f"Error en la entrada: {e}")
```
